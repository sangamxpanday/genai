```python
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
docs = [
    Document(page_content = "My name is Sangam panday"),
    Document(page_content = "Sangam panday is a student of langchain course"),
    Document(page_content = "Luffy is gonna be king of pirates" ),
    Document(page_content = "Gol D. Roger is the king of pirates" ),
    Document(page_content = "Naruto is a ninja" )
]
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")

vector_store = FAISS.from_documents(
    documents = docs,
    embedding = embeddings)
```

```python
retriever = vector_store.as_retriever(
    search_type="mmr",#Enables Maximal Marginal Relevance (MMR) search, which aims to balance relevance and diversity in the retrieved results.
    search_kwargs={"k": 2, "lambda_mult": 0.5})#lamda_mult = relevence-diversity balance
```

```python
query = "Who is the king of pirates?"
result = retriever.invoke(query)
```

```python
for i, doc in enumerate(result):
    print(f"Result {i+1}: {doc.page_content}")
```
